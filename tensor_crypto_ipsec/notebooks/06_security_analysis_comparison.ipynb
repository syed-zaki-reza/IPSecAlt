```python
print("\n📊 NIST SP 800-22 STATISTICAL TEST SUITE")
print("=" * 41)

print("\nRunning comprehensive randomness tests on tensor-generated keys...")

nist_results = crypto_analyzer.run_nist_tests()

print("\nNIST Test Results:")
for test_name, result in nist_results.items():
    status = "✅" if result['passed'] else "❌"
    print(f"• {test_name}: {result['status']} (p={result['p_value']:.4f}) {status}")

passed_tests = sum(1 for result in nist_results.values() if result['passed'])
total_tests = len(nist_results)

print(f"\n🎯 NIST Compliance: {passed_tests}/{total_tests} Tests PASSED - Exceeds Standards")
```
