```python
# 02_ai_logic_generation.ipynb
# AI-Generated Logic for Dynamic Cryptographic Operations
# Neural Network-based dynamic transformation tensor generation

import numpy as np
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import hashlib
import time
import matplotlib.pyplot as plt
from sklearn.metrics import mutual_info_score
import warnings
warnings.filterwarnings('ignore')

print("AI Logic Generation for Tensor-Based Cryptography")
print("=" * 50)

# Set seeds for reproducibility in demo
np.random.seed(42)
tf.random.set_seed(42)

# --- 1. AI MODEL ARCHITECTURE ---

print("\n1. AI MODEL ARCHITECTURE")
print("-" * 25)

# Define model dimensions
INPUT_DIM = 64   # Input seed vector size
OUTPUT_DIM = 16  # Output logic vector (reshape to 4x4 tensor)
HIDDEN_LAYERS = [128, 64, 32]

def build_ai_logic_model(input_dim=INPUT_DIM, output_dim=OUTPUT_DIM):
    """
    Build neural network for generating cryptographic logic
    Uses multiple hidden layers with different activations for complexity
    """
    model = keras.Sequential([
        # Input layer
        layers.Input(shape=(input_dim,), name='input_seed'),
        
        # Encoder layers - compress and transform input
        layers.Dense(HIDDEN_LAYERS[0], activation='relu', name='encoder_1'),
        layers.BatchNormalization(name='bn_1'),
        layers.Dropout(0.2, name='dropout_1'),
        
        layers.Dense(HIDDEN_LAYERS[1], activation='tanh', name='encoder_2'),
        layers.BatchNormalization(name='bn_2'),
        layers.Dropout(0.1, name='dropout_2'),
        
        # Bottleneck layer - critical transformation point
        layers.Dense(HIDDEN_LAYERS[2], activation='sigmoid', name='bottleneck'),
        
        # Decoder layers - expand to output
        layers.Dense(HIDDEN_LAYERS[1], activation='relu', name='decoder_1'),
        layers.BatchNormalization(name='bn_3'),
        
        layers.Dense(HIDDEN_LAYERS[0], activation='tanh', name='decoder_2'),
        layers.BatchNormalization(name='bn_4'),
        
        # Output layer - no activation for maximum range
        layers.Dense(output_dim, activation='linear', name='output_logic')
    ])
    
    # Custom loss function for cryptographic properties
    def crypto_loss(y_true, y_pred):
        # Standard MSE loss
        mse = tf.reduce_mean(tf.square(y_true - y_pred))
        
        # Add entropy regularization - encourage diverse outputs
        entropy_reg = -tf.reduce_mean(tf.reduce_sum(y_pred * tf.math.log(tf.abs(y_pred) + 1e-8), axis=1))
        
        # Add variance regularization - encourage spread
        variance_reg = tf.reduce_mean(tf.math.reduce_variance(y_pred, axis=1))
        
        return mse - 0.01 * entropy_reg + 0.01 * variance_reg
    
    model.compile(
        optimizer=keras.optimizers.Adam(learning_rate=0.001),
        loss=crypto_loss,
        metrics=['mse', 'mae']
    )
    
    return model

# Build and display model
ai_model = build_ai_logic_model()
print("AI Model Architecture:")
ai_model.summary()

print(f"Total parameters: {ai_model.count_params():,}")
print(f"Input dimension: {INPUT_DIM}")
print(f"Output dimension: {OUTPUT_DIM}")

# --- 2. SEED GENERATION FUNCTIONS ---

print("\n2. SEED GENERATION")
print("-" * 18)

def create_session_seed(session_id, device_fingerprint, timestamp=None):
    """
    Create deterministic but unpredictable seed from session parameters
    """
    if timestamp is None:
        timestamp = time.time()
    
    # Combine all session information
    combo_string = f"{session_id}_{device_fingerprint}_{timestamp}"
    
    # Create multiple hash values for diversity
    sha256_hash = hashlib.sha256(combo_string.encode()).digest()
    sha1_hash = hashlib.sha1(combo_string.encode()).digest()
    md5_hash = hashlib.md5(combo_string.encode()).digest()
    
    # Combine hashes
    combined_hash = sha256_hash + sha1_hash + md5_hash
    
    # Convert to seed vector
    seed = np.frombuffer(combined_hash, dtype=np.uint8)[:INPUT_DIM].astype(np.float32)
    
    # Normalize to [-1, 1] range for neural network
    normalized_seed = (seed - 128) / 128.0
    
    return normalized_seed

def create_training_data(num_samples=1000):
    """
    Generate training data for the AI model
    Creates diverse input-output pairs for cryptographic logic generation
    """
    X_train = []
    y_train = []
    
    for i in range(num_samples):
        # Generate diverse session parameters
        session_id = f"training_session_{i}"
        device_fp = f"device_{np.random.randint(100000, 999999)}"
        timestamp = time.time() + np.random.uniform(-86400, 86400)  # ±1 day
        
        # Create input seed
        seed = create_session_seed(session_id, device_fp, timestamp)
        X_train.append(seed)
        
        # Create target output (random but structured)
        # In practice, this would be based on desired cryptographic properties
        target = np.random.uniform(-2, 2, OUTPUT_DIM)
        target = np.tanh(target)  # Apply nonlinearity
        y_train.append(target)
    
    return np.array(X_train), np.array(y_train)

# Generate example seeds
print("Example seed generation:")
session_id = f"demo_session_{int(time.time())}"
device_fp = "device_demo_123456789abcdef"

seed = create_session_seed(session_id, device_fp)
print(f"Session ID: {session_id}")
print(f"Device fingerprint: {device_fp}")
print(f"Generated seed shape: {seed.shape}")
print(f"Seed range: [{seed.min():.3f}, {seed.max():.3f}]")
print(f"Seed sample: [{', '.join([f'{x:.3f}' for x in seed[:8]])}...]")

# --- 3. MODEL TRAINING ---

print("\n3. MODEL TRAINING")
print("-" * 17)

# Generate training data
print("Generating training data...")
X_train, y_train = create_training_data(2000)
print(f"Training data shape: X={X_train.shape}, y={y_train.shape}")

# Train the model
print("Training AI model...")
history = ai_model.fit(
    X_train, y_train,
    epochs=50,
    batch_size=32,
    validation_split=0.2,
    verbose=1,
    callbacks=[
        keras.callbacks.EarlyStopping(patience=10, restore_best_weights=True),
        keras.callbacks.ReduceLROnPlateau(patience=5, factor=0.5)
    ]
)

print("Training completed!")

# Plot training history
plt.figure(figsize=(12, 4))

plt.subplot(1, 2, 1)
plt.plot(history.history['loss'], label='Training Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.title('Model Loss During Training')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.grid(True, alpha=0.3)

plt.subplot(1, 2, 2)
plt.plot(history.history['mse'], label='Training MSE')
plt.plot(history.history['val_mse'], label='Validation MSE')
plt.title('Mean Squared Error During Training')
plt.xlabel('Epoch')
plt.ylabel('MSE')
plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('ai_training_history.png', dpi=150, bbox_inches='tight')
plt.show()

# --- 4. AI LOGIC GENERATION ---

print("\n4. AI LOGIC GENERATION")
print("-" * 23)

def generate_ai_logic(session_id, device_fingerprint, timestamp=None):
    """
    Generate AI-based cryptographic logic for a session
    """
    # Create input seed
    seed = create_session_seed(session_id, device_fingerprint, timestamp)
    
    # Generate logic using trained model
    logic_vector = ai_model.predict(seed.reshape(1, -1), verbose=0)[0]
    
    # Post-process for cryptographic properties
    processed_logic = post_process_logic(logic_vector)
    
    # Reshape to tensor format
    logic_tensor = processed_logic.reshape(4, 4)
    
    return logic_tensor, logic_vector

def post_process_logic(logic_vector):
    """
    Apply post-processing to enhance cryptographic properties
    """
    # Apply nonlinear transformations
    processed = np.tanh(logic_vector * 2)
    processed = np.sin(processed * np.pi)
    
    # Ensure non-zero values (avoid weak encryption)
    processed = np.where(np.abs(processed) < 0.01, 
                        np.sign(processed) * 0.01, processed)
    
    # Normalize to desired range
    processed = (processed / np.max(np.abs(processed))) * 127
    
    # Convert to integer for XOR operations
    processed = processed.astype(np.int8)
    
    return processed

# Generate example AI logic
print("Generating AI logic examples:")
examples = []
for i in range(3):
    session = f"example_session_{i}"
    device = f"example_device_{i}"
    
    logic_tensor, logic_vector = generate_ai_logic(session, device)
    examples.append((session, device, logic_tensor, logic_vector))
    
    print(f"\nSession: {session}")
    print(f"Device: {device}")
    print(f"Logic tensor shape: {logic_tensor.shape}")
    print(f"Logic tensor:\n{logic_tensor}")
    print(f"Logic vector range: [{logic_vector.min():.3f}, {logic_vector.max():.3f}]")

# --- 5. LOGIC QUALITY ANALYSIS ---

print("\n5. AI LOGIC QUALITY ANALYSIS")
print("-" * 30)

def analyze_logic_quality(num_samples=100):
    """
    Analyze the quality and randomness of generated AI logic
    """
    logic_samples = []
    entropy_scores = []
    
    for i in range(num_samples):
        session = f"analysis_session_{i}"
        device = f"analysis_device_{i}"
        
        logic_tensor, logic_vector = generate_ai_logic(session, device)
        logic_samples.append(logic_vector)
        
        # Calculate entropy (approximation)
        hist, _ = np.histogram(logic_vector, bins=20)
        hist = hist + 1e-10  # Avoid log(0)
        prob = hist / np.sum(hist)
        entropy = -np.sum(prob * np.log2(prob))
        entropy_scores.append(entropy)
    
    logic_matrix = np.array(logic_samples)
    
    # Analysis metrics
    mean_entropy = np.mean(entropy_scores)
    std_entropy = np.std(entropy_scores)
    
    # Inter-sample correlation
    correlations = []
    for i in range(min(50, num_samples-1)):
        for j in range(i+1, min(50, num_samples)):
            corr = np.corrcoef(logic_matrix[i], logic_matrix[j])[0,1]
            if not np.isnan(corr):
                correlations.append(abs(corr))
    
    mean_correlation = np.mean(correlations) if correlations else 0
    
    # Uniqueness
    unique_vectors = len(set(tuple(v) for v in logic_matrix))
    uniqueness_ratio = unique_vectors / num_samples
    
    return {
        'mean_entropy': mean_entropy,
        'std_entropy': std_entropy,
        'mean_correlation': mean_correlation,
        'uniqueness_ratio': uniqueness_ratio,
        'logic_samples': logic_matrix[:10]  # First 10 samples
    }

# Perform quality analysis
print("Analyzing AI logic quality...")
quality_metrics = analyze_logic_quality(100)

print("Quality Analysis Results:")
print(f"Mean entropy: {quality_metrics['mean_entropy']:.3f} bits")
print(f"Entropy std dev: {quality_metrics['std_entropy']:.3f}")
print(f"Mean correlation: {quality_metrics['mean_correlation']:.4f}")
print(f"Uniqueness ratio: {quality_metrics['uniqueness_ratio']:.3f}")

# Quality assessment
print("\nQuality Assessment:")
entropy_good = quality_metrics['mean_entropy'] > 3.0
correlation_good = quality_metrics['mean_correlation'] < 0.1
uniqueness_good = quality_metrics['uniqueness_ratio'] > 0.95

print(f"Entropy: {'GOOD' if entropy_good else 'NEEDS IMPROVEMENT'}")
print(f"Correlation: {'GOOD' if correlation_good else 'NEEDS IMPROVEMENT'}")
print(f"Uniqueness: {'GOOD' if uniqueness_good else 'NEEDS IMPROVEMENT'}")

# --- 6. VISUALIZATION ---

print("\n6. AI LOGIC VISUALIZATION")
print("-" * 25)

# Visualize AI logic tensors
fig, axes = plt.subplots(2, 3, figsize=(15, 10))

for i, (session, device, logic_tensor, logic_vector) in enumerate(examples):
    row, col = divmod(i, 3)
    
    # Plot logic tensor as heatmap
    im = axes[row, col].imshow(logic_tensor, cmap='RdBu_r', vmin=-127, vmax=127)
    axes[row, col].set_title(f'AI Logic Tensor {i+1}')
    axes[row, col].set_xlabel('Column')
    axes[row, col].set_ylabel('Row')
    
    # Add values as text
    for x in range(4):
        for y in range(4):
            axes[row, col].text(y, x, f'{logic_tensor[x,y]:3d}', 
                              ha='center', va='center', fontsize=8)
    
    plt.colorbar(im, ax=axes[row, col])

# Plot logic vector distribution
axes[1, 0].remove()
axes[1, 1].remove() 
axes[1, 2].remove()

ax_dist = plt.subplot(2, 1, 2)
all_values = np.concatenate([example[3] for example in examples])
ax_dist.hist(all_values, bins=30, alpha=0.7, edgecolor='black')
ax_dist.set_title('Distribution of AI Logic Values')
ax_dist.set_xlabel('Logic Value')
ax_dist.set_ylabel('Frequency')
ax_dist.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('ai_logic_visualization.png', dpi=150, bbox_inches='tight')
plt.show()

# --- 7. INTEGRATION WITH TENSOR ENCRYPTION ---

print("\n7. INTEGRATION WITH TENSOR ENCRYPTION")
print("-" * 38)

def ai_enhanced_encryption(message, tensor_key, session_id, device_fp):
    """
    Encrypt message using both tensor key and AI-generated logic
    """
    # Prepare message tensor (from notebook 1)
    if len(message) > 16:
        message = message[:16]
    message = message.ljust(16, '\0')
    msg_array = np.array([ord(c) for c in message], dtype=np.uint8).reshape(4, 4)
    
    # Generate AI logic
    ai_logic_tensor, _ = generate_ai_logic(session_id, device_fp)
    
    # Multi-layer encryption
    # Layer 1: Standard tensor key XOR
    encrypted_l1 = np.bitwise_xor(msg_array, tensor_key.astype(np.uint8))
    
    # Layer 2: AI logic XOR
    encrypted_l2 = np.bitwise_xor(encrypted_l1, ai_logic_tensor.astype(np.uint8))
    
    # Layer 3: Combined transformation
    combined_key = np.bitwise_xor(tensor_key.astype(np.uint8), 
                                 ai_logic_tensor.astype(np.uint8))
    final_encrypted = np.bitwise_xor(encrypted_l2, combined_key)
    
    return final_encrypted, ai_logic_tensor

def ai_enhanced_decryption(encrypted_tensor, tensor_key, ai_logic_tensor):
    """
    Decrypt message using both tensor key and AI logic
    """
    # Reverse Layer 3
    combined_key = np.bitwise_xor(tensor_key.astype(np.uint8), 
                                 ai_logic_tensor.astype(np.uint8))
    decrypted_l1 = np.bitwise_xor(encrypted_tensor, combined_key)
    
    # Reverse Layer 2
    decrypted_l2 = np.bitwise_xor(decrypted_l1, ai_logic_tensor.astype(np.uint8))
    
    # Reverse Layer 1
    final_decrypted = np.bitwise_xor(decrypted_l2, tensor_key.astype(np.uint8))
    
    return final_decrypted

# Test AI-enhanced encryption
test_message = "AI Enhanced Test"
test_key = np.random.randint(0, 256, (4, 4), dtype=np.uint8)
test_session = "integration_test_session"
test_device = "integration_test_device"

print("Testing AI-enhanced encryption:")
print(f"Original message: '{test_message}'")

encrypted, ai_logic = ai_enhanced_encryption(test_message, test_key, test_session, test_device)
decrypted = ai_enhanced_decryption(encrypted, test_key, ai_logic)

decrypted_message = ''.join([chr(x) for x in decrypted.flatten() if x != 0])
success = decrypted_message.strip() == test_message.strip()

print(f"Encrypted tensor:\n{encrypted}")
print(f"AI logic tensor:\n{ai_logic}")
print(f"Decrypted message: '{decrypted_message}'")
print(f"AI-enhanced encryption: {'SUCCESS' if success else 'FAILED'}")

# --- 8. PERFORMANCE EVALUATION ---

print("\n8. PERFORMANCE EVALUATION")
print("-" * 25)

# Benchmark AI logic generation
num_tests = 1000
start_time = time.time()

for i in range(num_tests):
    test_session = f"perf_session_{i}"
    test_device = f"perf_device_{i}"
    logic_tensor, _ = generate_ai_logic(test_session, test_device)

ai_generation_time = time.time() - start_time

print(f"Performance Results ({num_tests} iterations):")
print(f"AI logic generation time: {ai_generation_time:.3f} seconds")
print(f"Operations per second: {num_tests/ai_generation_time:.1f}")
print(f"Average time per generation: {(ai_generation_time/num_tests)*1000:.2f} ms")

# Compare with random generation
start_time = time.time()
for i in range(num_tests):
    random_logic = np.random.randint(-127, 128, (4, 4), dtype=np.int8)
random_generation_time = time.time() - start_time

print(f"\nComparison with random generation:")
print(f"Random generation time: {random_generation_time:.3f} seconds")
print(f"AI overhead factor: {ai_generation_time/random_generation_time:.1f}x")

# --- 9. SUMMARY ---

print("\n" + "=" * 50)
print("AI LOGIC GENERATION - SUMMARY")
print("=" * 50)
print("✓ Neural network model for AI logic generation")
print("✓ Deterministic seed generation from session parameters")
print("✓ Model training with cryptographic loss function")
print("✓ Logic quality analysis and validation")
print("✓ Integration with tensor-based encryption")
print("✓ Performance benchmarking completed")
print("✓ Ready for integration with dictionary management")
print("=" * 50)

# Save model for later use
ai_model.save('ai_logic_model.h5')
print("\nAI model saved as 'ai_logic_model.h5'")
```
