```python
import numpy as np
import tensorflow as tf
```

```python
# Generate a random tensor key (4x4 matrix for demo)
tensor_key = np.random.randint(0, 256, size=(4, 4), dtype=np.uint8)
print("Tensor Key:")
print(tensor_key)

```

```python
# Example message (must fit the tensor shape for this demo)
message = "TestMsg12345678"  # 16 characters

# Convert message to ASCII values and reshape
msg_array = np.array([ord(c) for c in message], dtype=np.uint8).reshape(tensor_key.shape)
print("Message Array:")
print(msg_array)
```

```python
def encrypt_tensor(msg_array, tensor_key):
    return np.bitwise_xor(msg_array, tensor_key)

encrypted = encrypt_tensor(msg_array, tensor_key)
print("Encrypted Array:")
print(encrypted)
```

```python
def decrypt_tensor(encrypted_array, tensor_key):
    return np.bitwise_xor(encrypted_array, tensor_key)

decrypted = decrypt_tensor(encrypted, tensor_key)
print("Decrypted Array:")
print(decrypted)

# Convert back to string
decrypted_message = ''.join([chr(x) for x in decrypted.flatten()])
print("Decrypted Message:", decrypted_message)
```
